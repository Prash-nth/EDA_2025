- find no. of outliers
- get data with and without outliers

```python
Q1 = wage_data.quantile(0.25)  
Q3 = wage_data.quantile(0.75) 
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers = wage_data[(wage_data < lower_bound) | (wage_data > upper_bound)]
print("Number of outliers:", len(outliers))
```

```python
q1=np.percentile(wage_data,25)
q2=np.percentile(wage_data,50)
q3=np.percentile(wage_data,75)
iqr=q3-q1
lb=q1-1.5*iqr
up=q3+1.5*iqr
con1=wage_data<lb
con2=wage_data>up
con=con1|con2
len(visa_df[con])
```

```python
q1=np.percentile(wage_data,25)
q2=np.percentile(wage_data,50)
q3=np.percentile(wage_data,75)
iqr=q3-q1
lb=q1-1.5*iqr
up=q3+1.5*iqr
con1=wage_data>lb
con2=wage_data<up
con=con1&con2
non_outlier=visa_df[con]
len(visa_df[con])
```
