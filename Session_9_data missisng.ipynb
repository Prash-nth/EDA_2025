- pad

```python
df.fillna(method='pad') # fills based on vlaue below it..if not exists..thgen nan
```

```python
df.fillna(method='bfill')  # fills based on vlaue above it..if not exists..thgen nan
```

```python
df.fillna(method='ffill')  # fills based on vlaue below it..if not exists..thgen nan
```

```python

df.fillna(method='backfill')  # fills based on vlaue above it..if not exists..thgen nan
```

**KNN Imputer**

- k-nearest neighbours
- k= hyper parameter means user can change
- Idea:Instead of taking all sample savg..
  - first fix neighbours numbers as k
  - then using distance formula..find neighbours
  -then take average of thise sanpleto find missing value

```python
import numpy as np
from sklearn.impute import KNNImputer
X=[[1,2,np.nan],[3,np.nan,3],[np.nan,60,5],[8,8,7]]
pd.DataFrame(X)
```

```python
imputer=KNNImputer(n_neighbors=2,weights='distance')
imputer.fit_transform(X)
```
