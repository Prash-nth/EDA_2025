```python
uni=visa_df['continent'].unique().tolist()
l=[]
for i in uni:
    a=(visa_df['continent']==i)
    b=len(visa_df[a])
    l.append(b)
l
```

```python
continents=uni
continent_df=pd.DataFrame(zip(labels,l),columns=['contient','Frequency'])
continent_df.to_csv('continent_df.csv',index=False)
```

```python
count=[]
for i in cat[1:]:
    labels=visa_df[i].unique()
    for label in labels:
        a=visa_df[i]==label
        count.append(len(visa_df[a]))
        count
    df=pd.DataFrame(zip(labels,count),columns=['contient','Frequency'])
    df.to_csv(f'{i}_df.csv',index=False)
```

**Value counts**

```python
continent_vc=visa_df['continent'].value_counts()
continent_vc
```

```python
for i in cat[1:]:
    vc=visa_df[i].value_counts()
    keys=vc.keys()
    values=vc.values
    vc_df=pd.DataFrame(zip(keys,values),columns=['Labels','Count'])
    vc_df.to_csv(f'{i}_df.csv',index=False)
```
