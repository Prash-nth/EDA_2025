# Normalization

```python
visa_df=pd.read_csv(r"C:\Users\achin\Downloads\DataSets(Nareshit)\Visadataset.csv")
cat=visa_df.select_dtypes(include='object').columns
num=visa_df.select_dtypes(exclude='object').columns
```

```python
x=visa_df['prevailing_wage']
x_min=np.min(x)
x_max=np.max(x)
nr=x - x_min
dr=x_max - x_min
wage_norm=nr/dr
wage_norm
```

```python
visa_df['wage_norm']=nr/dr
```

```python
visa_df[['prevailing_wage','wage_norm']].idxmax()
```

```python
visa_df[['prevailing_wage','wage_norm']].idxmin()
```
