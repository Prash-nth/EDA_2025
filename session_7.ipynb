```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
visa_df=pd.read_csv(r"C:\Users\achin\Downloads\DataSets(Nareshit)\Visadataset.csv")
cat=visa_df.select_dtypes(include='object').columns
num=visa_df.select_dtypes(exclude='object').columns
```

```python
con1=visa_df['continent']=='Asia'
con2=visa_df['case_status']=='Certified'
con=con1 & con2
visa_df[con]
```

```python
con1=visa_df['continent']=='Asia'
con2=visa_df['case_status']=='Certified'
con3=visa_df['case_status']=='Denied'
certi_con=con1 & con2
deni_con=con1 & con3
visa_df[deni_con]
```

```python
labels=visa_df['continent'].unique()
certi_count,deni_count=[],[]
for i in labels:
    con1=visa_df['continent']==i
    con2=visa_df['case_status']=='Certified'
    con3=visa_df['case_status']=='Denied'
    certi_con=con1 & con2
    deni_con=con1 & con3
    certi_count.append(len(visa_df[certi_con]))
    deni_count.append(len(visa_df[deni_con]))
certi_count,deni_count
analysis_df=pd.DataFrame(zip(certi_count,deni_count),columns=('Certified','Denied'),index=(labels))
analysis_df
```

```python
analysis_df.plot(kind='bar')
```
