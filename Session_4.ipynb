```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
visa_df=pd.read_csv(r"C:\Users\achin\Downloads\DataSets(Nareshit)\Visadataset.csv")
cat=visa_df.select_dtypes(include='object').columns
num=visa_df.select_dtypes(exclude='object').columns
```

```python
import os
folder_name='new_freq_table'
try: 
    os.makedirs(folder_name)
except Exception as e:
    print(e)

cwd=os.getcwd()
nwd=os.path.join(cwd,folder_name)

for i in cat[1:]:
    vc=visa_df[i].value_counts()
    keys=vc.keys()
    values=vc.values
    vc_df=pd.DataFrame(zip(keys,values),columns=['Labels','Count'])
    
    file_name=f'{i}_df.csv'
    path=os.path.join(nwd,file_name)
    
    vc_df.to_csv(path,index=False)
```

```python
class_names=visa_df['continent'].value_counts().keys()
class_fre=visa_df['continent'].value_counts().values

colors=['Red','Blue','Yellow','Black','Purple','Orange']
plt.figure(figsize=(10,5))
plt.bar(class_names,
        class_fre,
        color=colors)
plt.xlabel('Class_names')
plt.ylabel('Class_frequency')
plt.title('Continent_Barchat')
plt.savefig('continent_Barchart.jpg')
plt.show()
```

```python
class_names=visa_df['case_status'].value_counts().keys()
class_fre=visa_df['case_status'].value_counts().values

colors=['Red','Blue']
plt.figure(figsize=(10,5))
plt.bar(class_names,
        class_fre,
        color=colors)
plt.xlabel('Class_name')
plt.ylabel('Class_frequency')
plt.title('Job_Barchart')
plt.savefig('continent_Barchart.jpg')
plt.show()
```
